```
import os
import datetime
import random


DEVICE_TYPES = ['Temperature_Sensor', 'Humidity_Sensor', 'Pressure_Sensor', 'Light_Sensor']
LOCATIONS = ['Lab1', 'Lab2', 'OfficeA', 'OfficeB', 'Warehouse']
READING_RANGES = {
    'Temperature_Sensor': (18.0, 30.0),   # Celsius
    'Humidity_Sensor': (30.0, 80.0),      # Percentage
    'Pressure_Sensor': (980.0, 1030.0),   # hPa
    'Light_Sensor': (100.0, 1000.0)       # Lux
}
raw_sensor_data = []
for i in range(1, 101):
    timestamp = (
        datetime.datetime.now() -
        datetime.timedelta(minutes=random.randint(0, 5000))
    ).strftime("%Y-%m-%d %H:%M:%S")

    # Generate unique device ID
    device_id = f"DEV{i:03d}"

    # Randomly select device type
    device_type = random.choice(DEVICE_TYPES)

    # Randomly select location
    location = random.choice(LOCATIONS)

    # Generate sensor value based on device type
    min_value, max_value = READING_RANGES[device_type]
    value = round(random.uniform(min_value, max_value), 2)

    # Store data as a tuple
    sensor_reading = (
        timestamp,
        device_id,
        device_type,
        location,
        value
    )

    # Add tuple to list
    raw_sensor_data.append(sensor_reading)

# Print all 100 sensor readings
print("\nAll 100 Sensor Readings:\n")

for index, reading in enumerate(raw_sensor_data, start=1):
    print(f"{index}. {reading}")

# Print total number of readings
print("\nTotal Sensor Readings:", len(raw_sensor_data))
```

```
# 1.2 Create a list named 'temperature_readings'

temperature_readings = []

for reading in raw_sensor_data:
    timestamp, device_id, device_type, location, value = reading

    if device_type == "Temperature_Sensor":
        temp_dict = {
            "timestamp": timestamp,
            "device_id": device_id,
            "location": location,
            "temperature_celsius": value
        }

        temperature_readings.append(temp_dict)

# Display all temperature readings
print("Temperature Readings:\n")

for temp in temperature_readings:
    print(temp)

print("\nTotal Temperature Readings:", len(temperature_readings))
```

```
# 1.3 Create a set named 'unique_locations'

unique_locations = set()

for reading in raw_sensor_data:
    unique_locations.add(reading[3])   # Location is at index 3

print("Unique Locations:")
print(unique_locations)

print("Total Unique Locations:", len(unique_locations))
```

```
# 1.4 Create a dictionary named 'device_stats'

device_stats = {}

# Initialize with large/small values
for device in DEVICE_TYPES:
    device_stats[device] = (float('inf'), float('-inf'))

# Find minimum and maximum values for each device type
for reading in raw_sensor_data:
    timestamp, device_id, device_type, location, value = reading

    min_value, max_value = device_stats[device_type]

    if value < min_value:
        min_value = value

    if value > max_value:
        max_value = value

    device_stats[device_type] = (min_value, max_value)

# Display the dictionary
print("Device Statistics:")

for device_type, stats in device_stats.items():
    print(device_type, ":", stats)
```

```
print("\n--- Task 2.1: Get Readings by Location ---")

# Function to get all readings for a specific location
def get_readings_by_location(readings_list, target_location):
    location_readings = []

    for reading in readings_list:
        if reading[3] == target_location:
            location_readings.append(reading)

    return location_readings

# Example
location = "Lab1"
result = get_readings_by_location(raw_sensor_data, location)

print(f"\nSensor Readings for {location}:\n")

for reading in result:
    print(reading)

print("\nTotal Readings:", len(result))
```

```
print("\n--- Task 2.2: Calculate Average Temperature ---")

# Function to calculate average temperature
def calculate_average_temperature(temp_data_list):
    if len(temp_data_list) == 0:
        return 0.0

    total = 0

    for reading in temp_data_list:
        total += reading["temperature_celsius"]

    average = total / len(temp_data_list)
    return average

# Call the function
avg_temperature = calculate_average_temperature(temperature_readings)

print("Average Temperature:", round(avg_temperature, 2), "°C")
```

```
print("\n--- Task 2.3: Generate Report Line ---")

# Function to generate a formatted report line
def generate_report_line(reading_dict):
    return f"[{reading_dict['timestamp']}] Device {reading_dict['device_id']} at {reading_dict['location']}: {reading_dict['temperature_celsius']}°C"

# Example: Generate report for the first temperature reading
if len(temperature_readings) > 0:
    report = generate_report_line(temperature_readings[0])
    print(report)
else:
    print("No temperature readings available.")
```
