```python
import numpy as np;
import pandas as pd;
df = pd.DataFrame({
    'day':[1,2,3,4,5,6],
    'soilmoisture':[72,68,62,55,48,40],
    'temperature':[27,29,31,33,35,37],
    'Humidity':[78,74,70,65,60,55],
    'SoilNitrogen':[48,46,43,39,35,30],
    'CropHealthScore':[94,90,84,76,65,52],
})
df
```

```python
d=df.copy()
for i in d:
    print(d[i])
```

```python
d=df.copy()
d["status"]=np.where(
    d['soilmoisture']<50,'low soilmoisture alert' , 'soil moisture normal'
)
print(d[['soilmoisture','status']])
```

```python
d=df.copy()
d["status"]=np.where(
    d['temperature']>35,'high temperature alert','temperature normal'
)
print(d[['temperature','status']])
```

```python
d=df.copy()
def crop_status(score):
   if score>80:
         return 'healthy'
   elif score >= 50 and score <=80:
         return 'UnderStress'
   else:
         return 'critical'

d["status"]=d['CropHealthScore'].apply(crop_status)
print(d[['CropHealthScore','status']])
```

```python
d=df.copy()
arr=np.array(d)
print(arr)
```
